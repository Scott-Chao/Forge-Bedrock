```python
import sys
from pathlib import Path

project_root = str(Path.cwd().parent)
sys.path.append(project_root)

import numpy as np
import matplotlib.pyplot as plt
from core.linalg.matrix import Matrix

n_samples = 50
slope = 2
intercept = 1
noise_level = 2

np.random.seed(42)
X_raw = np.random.rand(n_samples, 1) * 10
noise = np.random.randn(n_samples, 1) * noise_level
y_raw = slope * X_raw + intercept + noise

X_data = np.hstack([X_raw, np.ones((n_samples, 1))])
X = Matrix(X_data)
y = Matrix(y_raw)

print(f"Dataset shape: X={X.shape}, y={y.shape}")
```

```python
# X.T @ X @ w = X.T @ y

# A. w = X.pinv @ y
w_stable = X.pinv @ y

# B. w = (X.T @ X).inv @ X.T @ y
XTX = X.T @ X
w_classic = XTX.inv @ X.T @ y

print("Weights (Stable SVD):", w_stable.data.flatten())
print("Weights (Classic Inv):", w_classic.data.flatten())
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X_raw, y_raw, color="blue", alpha=0.6, label="Observed Data")

x_test = np.linspace(0, 10, 100).reshape(-1, 1)
X_test_aug = Matrix(np.hstack([x_test, np.ones((100, 1))]))
y_pred = X_test_aug @ w_stable

plt.plot(x_test, y_pred.data, color="red", linewidth=2, label="Linear Fit")
plt.title("Least Squares Regression via Normal Equation")
plt.xlabel("X (Feature)")
plt.ylabel("y (Target)")
plt.legend()
plt.grid(True)
plt.show()
```
